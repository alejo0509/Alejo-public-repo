## Question 1
Write a function called `proportion_of_education` which returns the proportion of children in the dataset who had a mother with the education levels equal to less than high school (<12), high school (12), more than high school but not a college graduate (>12) and college degree.

*This function should return a dictionary in the form of (use the correct numbers, do not round numbers):*

````python
def proportion_of_education():
    proportions = df['EDUC1'].value_counts(normalize=True)

    result = {
        "less than high school": proportions.get(1),
        "high school": proportions.get(2),
        "more than high school but not college": proportions.get(3),
        "college": proportions.get(4)
    }
    return result
    raise NotImplementedError()
````

## Question 3
It would be interesting to see if there is any evidence of a link between vaccine effectiveness and sex of the child. Calculate the ratio of the number of children who contracted chickenpox but were vaccinated against it (at least one varicella dose) versus those who were vaccinated but did not contract chicken pox. Return results by sex. 

*This function should return a dictionary in the form of (use the correct numbers):* 
```
    {"male":0.2,
    "female":0.4}
```

Note: To aid in verification, the `chickenpox_by_sex()['female']` value the autograder is looking for starts with the digits `0.0077`.

- 1: Yes

P_NUMVRC – total number of varicella doses 

````python
#Posible values
np.sort(df['P_NUMVRC'].unique())
````

````python
vaccinated_male = df[
            (df["SEX"] == 1) &
            (df["P_NUMVRC"] >= 1)
        ]
vaccinated_male
````

````python
(vaccinated_male["HAD_CPOX"] == 1).sum()
````

````python
def chickenpox_by_sex():
    # Create an empty dictionary to store the results for each sex
    result = {}

    # Iterate over both sexes:
    # SEX = 1 corresponds to males and SEX = 2 corresponds to females
    for sex, name in [(1, "male"), (2, "female")]:

        # Filter the DataFrame to select only:
        # 1. Children of the current sex
        # 2. Children who received at least one dose of the varicella vaccine
        vaccinated = df[
            (df["SEX"] == sex) &
            (df["P_NUMVRC"] >= 1)
        ]

        # Count vaccinated children who contracted chickenpox
        # HAD_CPOX = 1 means the child had chickenpox
        had_chickenpox = (vaccinated["HAD_CPOX"] == 1).sum()

        # Count vaccinated children who did not contract chickenpox
        # HAD_CPOX = 2 means the child did not have chickenpox
        no_chickenpox = (vaccinated["HAD_CPOX"] == 2).sum()

        # Calculate the ratio of vaccinated children who had chickenpox
        # to vaccinated children who did not have chickenpox
        # Store the result under the "male" or "female" key
        result[name] = had_chickenpox / no_chickenpox

    # Return the dictionary containing the results for both sexes
    return result

    raise NotImplementedError()
````

````python
assert len(chickenpox_by_sex())==2, "Return a dictionary with two items, the first for males and the second for females."
````

## Question 4
A correlation is a statistical relationship between two variables. If we wanted to know if vaccines work, we might look at the correlation between the use of the vaccine and whether it results in prevention of the infection or disease [1]. In this question, you are to see if there is a correlation between having had the chicken pox and the number of chickenpox vaccine doses given (varicella).

Some notes on interpreting the answer. The `had_chickenpox_column` is either `1` (for yes) or `2` (for no), and the `num_chickenpox_vaccine_column` is the number of doses a child has been given of the varicella vaccine. A positive correlation (e.g., `corr > 0`) means that an increase in `had_chickenpox_column` (which means more no’s) would also increase the values of `num_chickenpox_vaccine_column` (which means more doses of vaccine). If there is a negative correlation (e.g., `corr < 0`), it indicates that having had chickenpox is related to an increase in the number of vaccine doses.

Also, `pval` is the probability that we observe a correlation between `had_chickenpox_column` and `num_chickenpox_vaccine_column` which is greater than or equal to a particular value occurred by chance. A small `pval` means that the observed correlation is highly unlikely to occur by chance. In this case, `pval` should be very small (will end in `e-18` indicating a very small number).

````python
def corr_chickenpox():
    import scipy.stats as stats
    import numpy as np
    import pandas as pd
    
    # Select the two columns needed for the correlation
    New_df = df[["HAD_CPOX", "P_NUMVRC"]].copy()

    # Keep only valid responses and remove missing values
    New_df = New_df[
        New_df["HAD_CPOX"].isin([1, 2]) &
        New_df["P_NUMVRC"].isin([0, 1, 2, 3])
    ].dropna()

    # Calculate the Pearson correlation coefficient
    corr, pval = stats.pearsonr(
        New_df["HAD_CPOX"],
        New_df["P_NUMVRC"]
    )

    return float(corr)
    
    raise NotImplementedError()
````
